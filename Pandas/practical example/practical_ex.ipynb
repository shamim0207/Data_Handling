```python
#selecting data
df["city"] # for single column
df[["city","aqi"]] # for miltiple column
df.loc[1] # for a single row wise data
df.loc[1:4] # for multuple row wise data
df.iloc[2] # using indexing
df.iloc[2:5]
df.loc[0,"aqi"] # for selecting row and column
```

```python
#filtering of data
df[df["aqi"]>100]
df[(df["aqi"] > 100) & (df["city"] == "New York")]
df[(df["aqi"] > 100) & (df["city"] == "New York") & (df["temperature"]>30)]
df[df["aqi"]>100][["city","aqi"]] # filtering specific column based on condition
df[(df["aqi"]>100) & (df["temperature"]>30)][["country","city","aqi","temperature"]]
```

```python
#data filtering using query
df.query("aqi > 100")
df.query("aqi > 100 & temperature > 30 & humidity > 80 ") [["city","aqi","temperature","humidity"]]
```
