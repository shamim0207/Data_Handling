```python
import json
```

```python
#load the data 
def load_data(filename):
    with open(filename,"r") as f:
        data= json.load(f)

    return data
```

```python
my_data = load_data("data.json")
print(my_data)
```

```python
# clean & structured the data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # clean rating 
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]  
        
        user["rating"] = raw_rating

        # handling missing values
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = None

        # de-duplication
        user_name = user["name"].strip()
        if user_name in unique_users:
            continue  
        unique_users.add(user_name)
        
        cleaned_data.append(user)

    return cleaned_data
```

```python
data = load_data("data.json")
clean_data(data)
```

```python
# get meaningful insight from data
def get_insight(data):
    if not data:
        print("No data available.")
        return

    # average rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    
    avg = total_rating / len(data)
    print(f"Average rating: {avg}")

    # percentage of users with poor rating (rating < 3)
    poor_rating = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_rating += 1
            
    percentage = (poor_rating / len(data)) * 100
    print(f"Poor rating percentage: {percentage}%")
```

```python
get_insight(data)
```

```python
# build recommendation feature (rule based no ML)
def get_recommendations(data):
    recommendations = []
    
    for user in data:
        curr_recommendation = {}
        

        curr_recommendation["name"] = user["name"]
        
        
        if float(user["rating"]) >= 4:
            curr_recommendation["recommended_product"] = "Apple"
        else:
            curr_recommendation["recommended_product"] = "Samsung"
            
        recommendations.append(curr_recommendation)
        
    return recommendations
```

```python
get_recommendations(data)
```
