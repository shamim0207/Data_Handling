```python
import pandas as pd
```

```python
#load the data file
df = pd.read_csv("raw_data.csv")
df
```

```python
# finiding missing value
df.isnull() # where null value find it's return True
```

```python
# find how many null value which column haveing
df.isnull().sum()
```

```python
# fill null value with a specific value for all null value in the dataset
df.fillna(45)
```

```python
# fill null value with a specific value for a specific column
df.fillna(35)["age"]
```

```python
# fill null value with a mean value for a specific column
age_mean = df["age"].mean()
print("age mean: ",age_mean)
df.fillna(age_mean)["age"]
```
